# 执行协程

Jupyter 自身就是一个持续运行的异步环境：它的内核（ipykernel）底层已经跑着一个 Event Loop 了。
如果在写`asyncio.run()`会报错
RuntimeError: asyncio.run() cannot be called from a running event loop

```python
import asyncio

from loguru import logger


async def hi_world():
    logger.info("hi start")
    await asyncio.sleep(1)
    logger.info("world end")


# asyncio.run(hi_world())
await hi_world()
```

# 串行协程

```python
import time


async def work(name, delay):
    await asyncio.sleep(delay)
    logger.info(f"{name}执行完成")


async def serial_run():
    logger.info("START")
    start_time = time.perf_counter()
    await work("Task A", 3)
    await work("Task B", 2)
    end_time = time.perf_counter()
    logger.info(f"END {(end_time - start_time):.2f}")
```

```python
await serial_run()
```
