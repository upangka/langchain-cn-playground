# LangChain访问硅基流动的模型

1. 硅基流动平台
2. LangChain官网
3. **目的**：搭建好基础环境

建立[LangChain](https://docs.langchain.com/oss/python/langchain/models)连接[硅基流动](https://cloud.siliconflow.cn/me/models)的`deepseek-ai/DeepSeek-V4.1-Flash`模型

但是这些模型平台其实我不太信任，还是官网直连比较好。

```python
import os

import dotenv

dotenv.load_dotenv()
os.environ["base_url"]
```

通过具体的[langchain models api](https://docs.langchain.com/oss/python/langchain/models)

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model=os.environ["model_name"],
    base_url=os.environ["base_url"],
    api_key=os.environ["api_key"],
)
```

```python
resp = model.invoke("你好，我是使用langchain通过硅基流动平台上的模型访问，请问你是谁？")
```

```python
from pprint import pprint

pprint(resp)
```

```python
type(resp)
```

```python
print(resp.content)
```

## create_agents

另一种API

```python
from langchain.agents import create_agent

agent = create_agent(model=model)
```

```python
agent_resp = agent.invoke(
    {"messages": [{"role": "user", "content": "现在你所知道的deepseek模型是什么?"}]},
)
```

```python
type(agent_resp)
```

```python
agent_resp
```

```python
pprint(agent_resp["messages"][-1].content_blocks)
```
